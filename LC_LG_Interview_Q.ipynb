**1. What is LangChain, and what problems does it solve compared with calling an LLM API directly?**

`A`.LangChain is an open-source framework for building applications powered by large language models. Compared with calling an LLM API directly, it provides reusable abstractions and integrations for prompt templates, model providers, document processing, retrieval, tool calling, and structured outputs. Direct API calls are often sufficient for simple applications, while LangChain becomes useful when an application needs multiple components or complex processing pipelines. However, LangChain is an abstraction layer, not a replacement for the underlying LLM API.
